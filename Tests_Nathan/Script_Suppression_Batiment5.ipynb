```python
fds = pd.read_csv("../DataCleaning/full_dataset_clean.csv")

# Liste des colonnes à garder
colonnes_a_garder = [
    'production', 'p_per_m2', 'panel_area', 'facade_area',
    'total_panel_area', 'air_temp_mean', 'air_temp_std',
    'cloud_opacity_mean', 'cloud_opacity_min', 'cloud_opacity_q1',
    'cloud_opacity_q2', 'cloud_opacity_q3', 'cloud_opacity_max', 'dni_mean',
    'dni_std', 'dni_q3', 'dni_max', 'ghi_mean', 'ghi_std', 'ghi_q3',
    'ghi_max', 'gti_mean', 'gti_std', 'gti_q3', 'gti_max',
    'relative_humidity_mean', 'relative_humidity_std',
    'relative_humidity_min', 'relative_humidity_q1', 'building_id'
]

# Filtrage du DataFrame
fds = fds[colonnes_a_garder]
```

```python
#supression des lignes avec building_id = 5
fds = fds[fds['building_id'] != 5]
```

```python
#Drop building_id column
fds = fds.drop(columns=['building_id'])
# Save the filtered DataFrame to a new CSV file
fds.to_csv("../DataCleaning/scaled_dataset_clean_sansBatiment5.csv", index=False)
```
